```python
def polar_string (c_number):
    return (f'{cmath.polar(c_number)[0]}∠{np.degrees(cmath.polar(c_number)[1])}°')
```

```python
xd = 1.182
xq = 0.416
delta_max = np.deg2rad(69)
```

```python
def calc_ef(Vt, xd, xq, delta_max):
    termo1 = 4*Vt*(xd - xq) / (2 * np.cos(-delta_max) * xq)
    termo2 = 0.5 - np.power(np.cos(-delta_max), 2)
    return termo1 * termo2
```

```python
Ef = calc_ef(Vt=1, xd=xd, xq=xq, delta_max=delta_max)
print(f'Ef: {Ef} pu.')
```

```python
def P_polos_salientes(Ef, Vt, xd, delta, xq):
    P_eletromag=  Ef *Vt / xd * np.sin(-delta)
    P_relut=  Vt*Vt*(xd-xq)/(2*xd*xq)*np.sin(-2*delta)
    return {
        'P_eletromag': P_eletromag,
        'P_relut': P_relut
    }
```

```python
Pb_dict = P_polos_salientes(Ef=Ef, Vt=1, xd = xd, delta = delta_max, xq=xq)
Pb = Pb_dict['P_eletromag'] + Pb_dict['P_relut']
print(Pb_dict)
print(f'P total = {Pb}')
```

```python
Pc = P_polos_salientes(Ef = Ef, Vt = 1, xd = xd, xq = xq, delta = np.deg2rad(-45)) #aqui o Ef não importa
print(f'Prelut = {Pc["P_relut"]}')
```

```python
Id = cmath.rect(  np.cos(np.deg2rad(-45)) / xd ,
                np.deg2rad(-135))


Iq = cmath.rect(-1*np.sin(np.deg2rad(-45)) / xq, np.deg2rad(-45))
Ia = Id + Iq

print(f'Id: {polar_string(Id)}')
print(f'Iq: {polar_string(Iq)}')
print(f'Ia: {polar_string(Ia)}')
```

```python
Vl = 725 #tensao de linha
fp = 0.45
P3 = 35000 * 0.8#potencia trifásica
f = 60
polos = 2
xd = 13.4717
xq = 6.08223
```

```python
Ia_mag = P3 / fp / (np.sqrt(3) * Vl)
Ia_ang = -np.acos(fp)
Ia = cmath.rect(Ia_mag, Ia_ang)
print(f'Ia: {polar_string(Ia)}')
```

```python
Ef_duas_linhas = Vl/np.sqrt(3) - Ia * xq * (0 + 1j)
delta = cmath.phase(Ef_duas_linhas)
polar_string(Ef_duas_linhas)
```

```python
phi = np.acos(fp)
Iq_mag = Ia_mag * np.cos(delta + phi)
Iq = cmath.rect(Iq_mag, delta)
polar_string(Iq)
```

```python
Id_mag = np.abs(Ia_mag * np.sin(delta + phi))
Id = cmath.rect(Id_mag, delta - np.pi/2)
polar_string(Id)
```

```python
Ef = Vl / np.sqrt(3) - Id * xd*(0+1j) - Iq * xq*(0+1j)
polar_string(Ef) 
```
