````python
def print_polar (c_number):
    print(f'{cmath.polar(c_number)[0]}∠{np.degrees(cmath.polar(c_number)[1])}°')
    
def polar_string (c_number):
    return (f'{cmath.polar(c_number)[0]}∠{np.degrees(cmath.polar(c_number)[1])}°')
````

````python
S_c = 0.8 * 225000
V_l = 1550
#calculo da corrente 
I_a_mag = S_c / (V_l * np.sqrt(3))
fp = 0.85
phi_rad = np.arccos(0.85)
I_a = cmath.rect(I_a_mag, -phi_rad)
print(f'Ia: {polar_string(I_a)}')
````

A função ```np.interp()``` será usada para realizar as intepolações lineareas entre os pontos da tabela da lista para resolver as alternativas que exigem interpolação.

````python
#Do circuito equivalente:
E_r = (V_l / np.sqrt(3))+ (x_a * (0 + 1j)) *I_a
print(f'Er: {polar_string(E_r)}')
````

````python
#Interpolacao para encontrar Ifr
E_r_mag = np.absolute(E_r)

If_cca = np.array([
    0.000000, 448.074, 597.432, 746.790, 896.148, 1045.51, 1135.12,
    1194.86, 1344.22, 1418.90, 1493.58, 1642.94, 1792.30, 1941.65,
    2091.01, 2240.37, 2389.73, 2539.09, 2688.44
])

Efs_cca = np.array([
    0.000000, 467.999, 620.839, 735.010, 817.165, 878.191, 908.568,
    927.018, 967.846, 985.366, 1000.78, 1025.87, 1050.61, 1069.33,
    1083.93, 1095.17, 1104.16, 1110.89, 1117.63
])

I_fr_mag = np.interp(E_r_mag, Efs_cca, If_cca)
I_fr_ang = np.deg2rad(90) + cmath.phase(E_r)
I_fr = cmath.rect(I_fr_mag,  I_fr_ang)
print(f'I_fr: {polar_string(I_fr)}')
````

````python
#obtendo Ifa:
I_fa = K_a * I_a
print(f'I_fa: {polar_string(I_fa)}')
````

````python
#obtendo If:
I_f = I_fr - I_fa
print(f'I_f: {polar_string(I_f)}')
````

````python
#Interpolacao para encontrar Ef
I_f_mag = np.absolute(I_f)
E_f_mag = np.interp(I_f_mag, If_cca, Efs_cca)
E_f_ang = np.angle(I_f) - np.deg2rad(90)
E_f = cmath.rect(E_f_mag, E_f_ang)
print(f'E_f: {polar_string(E_f)}')
````

````python
#calculo da regulacao:
Reg = (E_f_mag - V_l/np.sqrt(3)) / (V_l* np.sqrt(3))
print(f'Regulacao: {Reg * 100}%')
````
