Começando as funções para cálculo das potências

````python
def P_polos_salientes(Ef, Vt, xd, delta, xq):
    P_eletromag=3 * Ef *Vt / xd * np.sin(delta)
    P_relut=3 * Vt*Vt*(xd-xq)/(2*xd*xq)*np.sin(2*delta)
    return {
        'P_eletromag': P_eletromag,
        'P_relut': P_relut
    }
````

````python
def Q_polos_salientes(Ef, Vt, xd, delta, xq):
    termo1=Ef*Vt/xd*np.cos(delta)
    termo2=Vt*Vt/xd
    termo3=Vt*Vt*(xd-xq)/(xd*xq)*np.power(np.sin(delta),2)
    return 3*(termo1 - termo2 - termo3) 
````

````python
def Q_polos_lisos(Ef, Vt, xs, delta):
    termo1 = Ef*Vt/xs * np.cos(delta)
    termo2 = Vt*Vt / xs
    return 3 * (termo1 - termo2)
````

````python
def P_polos_lisos(Ef ,Vt, xs, delta):
    return 3 * Ef * Vt / xs * np.sin(delta)
````
