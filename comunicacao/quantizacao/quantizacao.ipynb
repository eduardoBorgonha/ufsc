# Quantização
Notebook para aplicar diferentes níveis de quantização à um áudio.

```python
import numpy as np
import soundfile as sf
import os
```

Primeiramente, vamos criar a função de quantização do áudio:

```python
def quantize_midrise(signal, n_bits):
    n_levels = 2 ** n_bits  # número de níveis
    min_val, max_val = -1.0, 1.0
    delta = (max_val - min_val) / n_levels  # passo de quantização
    
    # aplica a quantização mid-rise
    quantized = np.floor((signal - min_val) / delta) * delta + (delta / 2) + min_val
    quantized = np.clip(quantized, -1, 1 - delta)  # saturação
    return quantized
```

Leitura do arquivo wav para array do numpy

```python
y, Fs = sf.read("spfc_tricampeao.wav")

# convertendo para mono:
if y.ndim > 1:
    y = np.mean(y, axis=1)
```

Agora, vamos definir quantos bits usaremos para a quantização e em seguida quantizar o áudio do São Paulo Futebol Clube sendo tricampeão mundial em vários níveis:

```python
n_bits_list = [2 ,4 ,8 ,16]
for n_bits in n_bits_list:
    print(f"Processando quantização com {n_bits} bits...")
    
    y_quantized = quantize_midrise(y, n_bits)
    
    # Normaliza novamente (evita clipping ao salvar)
    y_quantized = y_quantized / np.max(np.abs(y_quantized))
    
    # Gera nome do arquivo de saída
    base_name, ext = os.path.splitext("spfc_tricampeao.wav")
    output_file = f"{base_name}_quantizado_{n_bits}bits{ext}"
    
    # Salva o arquivo de áudio quantizado
    sf.write(output_file, y_quantized, Fs)
    print(f"--> Áudio salvo como: {output_file}")
```
